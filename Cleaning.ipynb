```python
import pandas as pd
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import OneHotEncoder,StandardScaler

df = pd.read_csv("hf://datasets/nirm2002/wanderplan-dataset/wanderplan_dataset_final.csv")
df.head()
df.isnull().sum()

#isolaion forest

df_num=df[["base_daily_cost_usd","base_flight_cost_usd","rating","latitude", "longitude"]]

iso = IsolationForest( contamination=0.05, random_state=42)
Anomalies=iso.fit_predict(df_num)

df_num["Anomalies"]=pd.DataFrame(Anomalies)

df_num[df_num["Anomalies"]==-1].value_counts().sum()

DATA=pd.concat([df_num["Anomalies"],df],axis=1)
DATA=DATA[DATA["Anomalies"]==1]
DATA=DATA.drop(["Anomalies","destination_id","detailed_description"],axis=1)
DATA.value_counts().sum()
DATA.head()
#codage
categ_nam=["city_name","category","destination_continent","country","target_demographic","best_travel_season"]

df_cat = DATA[categ_nam]
df_num = DATA.drop(columns=categ_nam)

encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

encoded = encoder.fit_transform(df_cat)

df_encoded = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(categ_nam),
    index=DATA.index
)

DATA= pd.concat([df_num, df_encoded], axis=1)

#stand
scaler = StandardScaler()
DATA[df_num.columns] = scaler.fit_transform(DATA[df_num.columns])
DATA = DATA.drop_duplicates(keep="first").reset_index(drop=True)
DATA.head()
DATA.shape
DATA.to_csv("data_withOut_Outliers/DATA_without_Outliers.csv",index=False)
DATA.shape
```

```python

import pandas as pd
from sklearn.preprocessing import OneHotEncoder,StandardScaler

df = pd.read_csv("hf://datasets/nirm2002/wanderplan-dataset/wanderplan_dataset_final.csv")
df.head()
df.isnull().sum()

#preprocessing

DATA = df.drop(["destination_id","detailed_description"],axis=1)

DATA = DATA.drop_duplicates(keep="first").reset_index(drop=True)

DATA.head()
DATA.shape

#codage

categ_nam=["city_name","category","destination_continent","country","target_demographic","best_travel_season"]

df_cat = DATA[categ_nam]
df_num = DATA.drop(columns=categ_nam)

encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

encoded = encoder.fit_transform(df_cat)

df_encoded = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(categ_nam),
    index=DATA.index
)

DATA= pd.concat([df_num, df_encoded], axis=1)

#stand

scaler = StandardScaler()
DATA[df_num.columns] = scaler.fit_transform(DATA[df_num.columns])

DATA = DATA.drop_duplicates(keep="first").reset_index(drop=True)

DATA.head()


DATA.to_csv("data_with_Outliers/DATA_with_Outliers.csv",index=False)
DATA.shape
```
